## XML คืออะไร
- ย่อมาจาก eXtensible Markup Language
- เป็นภาษา Markup คล้ายกันกับ HTML
- ถูกออกแบบมาเพื่อจัดเก็บและส่งผ่านข้อมูล
- ออกแบบมาเพื่อให้สามารถอธิบายข้อมูลได้ด้วยตัวเอง

```xml
<note>
  <to>Tove</to>
  <from>Jani</from>
  <heading>Reminder</heading>
  <body>Don't forget me this weekend!</body>
</note>
```

### XML ตัวอย่างด้านบนสามารถอธิบายตัวเองได้ดีมาก
เพราะว่ามันมีข้อมูลครบถ้วนดังนี้
1. ข้อมูลผู้รับ
2. ข้อมูลผู้ส่ง
3. หัวเรื่อง
4. เนื้อหาของข้อความ

แต่มันก็มีไว้แค่เก็บข้อมูลไม่ได้เอาไปใช้สั่งการให้มันทำอะไร

### ความแตกต่างระหว่าง XML และ HTML
มันถูกออกแบบมาให้มีเป้าหมายที่แตกต่างกัน
- XML ออกแบบมาเพื่อนำส่งข้อมูล โดยเน้นว่าข้อมูลนั้นน่าตาเป็นอย่างไร
- Tag ของ XML ไม่ได้ถูกกำหนดไว้ล่วงหน้าเหมือนกับ HTML

### ภาษา XML ไม่ใช้แท็กที่ถูกกำหนดไว้ล่วงหน้า
ภาษา XML ไม่ได้มีแท็กมาตราฐานที่กำหนดมาให้ ขึ้นอยู่กับผู้ใช้เขียนเลย

### XML are Extensible
ภาษา XML สามารถขยายต่อได้เรื่อย ๆ โดยสามารถเพิ่มเนื้อหาเข้าไปภายในเพิ่มเติมได้

```xml
<note>
  <to>Tove</to>
  <from>Jani</from>
  <heading>Reminder</heading>
  <body>Don't forget me this weekend!</body>
</note>
```

```xml
<note>
  <date>2015-09-01</date>
  <hour>08:30</hour>
  <to>Tove</to>
  <from>Jani</from>
  <body>Don't forget me this weekend!</body>
</note>
```

### ปกติ xmkl ใช้แยกข้อมูลกับส่วนแสดงผล
มักทำงานร่วมกับ HTML
โดยเรียก HTML มาเพื่อจัดรูปแบบหรือแสดงผลข้อมูลนั้นๆ

### XML Attributes คืออะไร
องค์ประกอบของ XML Elements โดย XML ก็สามาระมี Attributes ได้เหมือนกันกับ HTML โดยออกแบบมาไว้ใช้เก็บข้อมูลที่เกี่ยวข้องกับองค์ประกอบนั้น ๆ โดยเฉพาะ

### XML Elements vs Attributes

```xml
<person gender="female">
  <firstname>Anna</firstname>
  <lastname>Smith</lastname>
</person>
```

```xml
<person>
  <gender>female</gender>
  <firstname>Anna</firstname>
  <lastname>Smith</lastname>
</person>
```

## XPath คืออะไร
- ย่อมาจาก XML Path Language
- ใช้ไวยากรณ์แบบ "path like" เพื่อระบุและนำทางไปยังโหนดต่าง ๆ ในเอกสาร XML
- XPath มีฟังก์ชันในตัวมากกว่า 200 ฟังก์ชัน
- เป็นองค์ประกอบหลักในมาตราฐาน XSLT

### XPath ใช้การแสดงนิพจน์เส้นทางเพื่อเลือก Node หรือกลุ่ม Node ในเอกสาร XML
มันคล้ายคลึงกับนิพจน์ระบุเส้นทางที่ใช้กับระบบไฟล์คอมพิวเตอร์แบบดั้งเดิม

```markdown
Folder
- Books
    - Book
        - Title
        - Author
            - Firstname
            - Lastname
```

### XPath มีฟังก์ชันมาตราฐานในตัวกว่า 200 ฟังก์ชัน
- สำหรับส่งค่าสตริง ตัวเลข บูลีน
- การเปรียบเทียบวันที่และเวลา
- การจัดการ Node จัดการลำดับและอื่น ๆ
- ปัจจุบันนิพจน์ XPath สามารถนำไปใช้งานได้ในหลายภาษา เช่นพวก Javascripts, Java, XML Schema, PHP, Python, C, และ C++

### ศัพท์เฉพาะของ XPath
#### Node
ใน XPath จะมี Node 7 ประเภท
1. Element Node (องค์ประกอบ)
2. Attributes Node
3. Text Node (ข้อความ)
4. Namespace Node
5. Processsing-Instruction  Node (คำสั่งประมวลผล)
6. Comment Node (ความคิดเห็น)
7. Root Node

เอกสาร XML จะถูกมองว่าเป็นโครงสร้างแบบต้นไม้ที่มี Node ต่าง ๆ เรียงกัน โดยองค์ประกอบที่อยู่บนสุดของต้นไม้เรียกว่าองค์ประกอบราก

```xml
<?xml version="1.0" encoding="UTF-8"?>

<bookstore>
  <book>
    <title lang="en">Harry Potter</title>
    <author>J K. Rowling</author>
    <year>2005</year>
    <price>29.99</price>
  </book>
</bookstore>
```

```xml
<bookstore> (root element node)

<author>J K. Rowling</author> (element node)

lang="en" (attribute node)
```

```markdown
### ค่าอะตอม (Atomic Values)
คือค่าข้อมูลเดี่ยว ๆ ที่ไม่มีโครงสร้างภายในเป็น Node ย่อยอีกต่อไป เหมือนเป็น 'ใบ' ของโครงสร้างต้นไม้ XML

**ตัวอย่างจาก XML ด้านบน:**
- `J K. Rowling` (นี่คือข้อความที่อยู่ใน Element `<author>`) เป็น Atomic Value
- `en` (นี่คือค่าของ Attribute `lang` ใน `<title lang="en">`) เป็น Atomic Value
- `29.99` (นี่คือข้อความที่อยู่ใน Element `<price>`) เป็น Atomic Value

### รายการ (Items)
ใน XPath, 'รายการ' คือลำดับของค่า (Sequence of Values) ซึ่งอาจจะเป็น:
1. **Atomic Values:** เช่น `J K. Rowling` หรือ `29.99`
2. **Nodes:** เช่น Element Node ทั้งก้อนอย่าง `<author>J K. Rowling</author>` หรือ Attribute Node `lang="en"`

พูดง่าย ๆ คือ เมื่อ XPath เลือกสิ่งใดสิ่งหนึ่งออกมา ไม่ว่าจะเป็นค่าข้อมูลเดี่ยว ๆ หรือ Node ทั้งก้อน สิ่งที่ได้ออกมานั้นเรียกว่า 'รายการ' (Item) ซึ่งรายการนี้สามารถมีได้หลาย 'Item' ถ้ามีการเลือกหลายสิ่งพร้อมกันค่ะ
```

### Relationshio of Nodes
#### Parent
แต่ละองค์ประกอบและคุณลักษณะจะมีผู้ปกครองเพียงแค่หนึ่งเดียว

ในตัวอย่างด้านล่างนี้ องค์ประกอบหนังสือเป็นองค์ประกอบหลักของ ชื่อเรื่อง ผู้แต่ง ปี และ ราคา

```xml
<book>
  <title>Harry Potter</title>
  <author>J K. Rowling</author>
  <year>2005</year>
  <price>29.99</price>
</book>
```

#### Child (องค์ประกอบย่อย)
Node องค์ประกอบอาจจะไม่มีลูก มีหนึ่ง หรือมีมากกว่านั้น
ในตัวอย่างด้านล่างนี้ ชื่อเรื่อง ผู้แต่ง ปี ราคา ล้วนเป็นลูกหรือองค์ประกอบย่อยของหนังสือ

```xml
<book>
  <title>Harry Potter</title>
  <author>J K. Rowling</author>
  <year>2005</year>
  <price>29.99</price>
</book>
```

#### Siblings พี่น้อง
โหนดที่มีผู้ปกครองเดียวกัน

ในตัวอย่างต่อไปนี้ ชื่อเรื่อง ผู้แต่ง ปี และราคา ล้วนเป็นองค์ประกอบที่อยู่ในลำดับเดียวกัน:

```xml
<book>
  <title>Harry Potter</title>
  <author>J K. Rowling</author>
  <year>2005</year>
  <price>29.99</price>
</book>
```

#### Ancestor บรรพบุรุษ
พวก Node แม่ ของ Node แม่ (ยาย ยายทวดอะไรแบบนี้)
ในแต่ละตัวอย่างต่อไปนี้ บรรพบุรุษขององค์ประกอบชื่อเรื่องคือองค์ประกอบหนังสือ และองค์ประกอบร้านหนังสือ

```xml
<bookstore>

<book>
  <title>Harry Potter</title>
  <author>J K. Rowling</author>
  <year>2005</year>
  <price>29.99</price>
</book>

</bookstore>
```

#### Descendants ลูกหลาน
ลูกๆ ของโหนด ลูกๆ ของลูกๆ เป็นต้น

ในตัวอย่างต่อไปนี้ องค์ประกอบย่อยของร้านหนังสือ ได้แก่ หนังสือ ชื่อเรื่อง ผู้แต่ง ปี และราคา:

```xml
<bookstore>

<book>
  <title>Harry Potter</title>
  <author>J K. Rowling</author>
  <year>2005</year>
  <price>29.99</price>
</book>

</bookstore>
```

### การเลือก Node
XPath ใช้การแสดงเส้นทางเพื่อเลือกโหนดในเอกสาร XML โดยการเลือกโหนดทำได้โดยการติดตามเส้นทางหรือขั้นตอนต่าง ๆ การแสดงเส้นทางที่มีประโยชน์ที่สุดมีดังต่อไปนี้:

| Expression | Description |
|:-----------|:------------|
| `nodename` | เลือกโหนดทั้งหมดที่มีชื่อ "nodename" |
| `/`        | เลือกจากโหนด root |
| `//`       | เลือกโหนดในเอกสารจากโหนดปัจจุบันที่ตรงกับการเลือก ไม่ว่าจะอยู่ที่ใดก็ตาม |
| `.`        | เลือกโหนดปัจจุบัน |
| `..`       | เลือกโหนดแม่ของโหนดปัจจุบัน |
| `@`        | เลือก attributes |

ในตารางด้านล่างนี้ เราได้แสดงรายการนิพจน์เส้นทางบางส่วนและผลลัพธ์ของนิพจน์เหล่านั้น:

| Path Expression | Result |
|:----------------|:-------|
| `bookstore`     | เลือกโหนดทั้งหมดที่มีชื่อว่า "bookstore" |
| `/bookstore`    | เลือก root element ที่ชื่อ bookstore \n**หมายเหตุ:** หาก path เริ่มต้นด้วยเครื่องหมายทับ (`/`) จะแสดงถึง absolute path ไปยัง element เสมอ! |
| `bookstore/book` | เลือก element book ทั้งหมดที่เป็นลูกของ bookstore |
| `//book`        | เลือก element book ทั้งหมด ไม่ว่าจะอยู่ที่ใดในเอกสาร |
| `bookstore//book` | เลือก element book ทั้งหมดที่เป็นลูกหลานของ element bookstore ไม่ว่าจะอยู่ที่ใดภายใต้ element bookstore |
| `//@lang`       | เลือก attributes ทั้งหมดที่มีชื่อว่า lang |

### Predicates (เพรดิเคต)
Predicates ใช้เพื่อค้นหาโหนดที่เฉพาะเจาะจง หรือโหนดที่มีค่าเฉพาะเจาะจง

Predicates จะถูกฝังอยู่ในวงเล็บเหลี่ยมเสมอ `[]`

ในตารางด้านล่างนี้ เราได้แสดงรายการนิพจน์เส้นทางบางส่วนพร้อม Predicates และผลลัพธ์ของนิพจน์เหล่านั้น:

| Path Expression             | Result                                                                                                                                                                                                            |
|:----------------------------|:--------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------|
| `/bookstore/book[1]`        | เลือก `book` element ตัวแรกที่เป็นลูกของ `bookstore` element \n**หมายเหตุ:** ใน IE 5,6,7,8,9 โหนดแรกคือ `[0]` แต่ตาม W3C คือ `[1]` หากต้องการแก้ไขปัญหานี้ใน IE ให้ตั้งค่า SelectionLanguage เป็น XPath: \n`In JavaScript: xml.setProperty("SelectionLanguage","XPath");` |
| `/bookstore/book[last()]`   | เลือก `book` element ตัวสุดท้ายที่เป็นลูกของ `bookstore` element                                                                                                                                             |
| `/bookstore/book[last()-1]` | เลือก `book` element ตัวรองสุดท้ายที่เป็นลูกของ `bookstore` element                                                                                                                                         |
| `/bookstore/book[position()<3]` | เลือก `book` element สองตัวแรกที่เป็นลูกของ `bookstore` element                                                                                                                                                |
| `//title[@lang]`            | เลือก `title` element ทั้งหมดที่มี attribute ชื่อ `lang`                                                                                                                                                    |
| `//title[@lang='en']`       | เลือก `title` element ทั้งหมดที่มี attribute `lang` ซึ่งมีค่าเป็น "en"                                                                                                                                        |
| `/bookstore/book[price>35.00]` | เลือก `book` element ทั้งหมดของ `bookstore` element ที่มี `price` element ที่มีค่ามากกว่า 35.00                                                                                                     |
| `/bookstore/book[price>35.00]/title` | เลือก `title` element ทั้งหมดของ `book` element ของ `bookstore` element ที่มี `price` element ที่มีค่ามากกว่า 35.00                                                                                             |
