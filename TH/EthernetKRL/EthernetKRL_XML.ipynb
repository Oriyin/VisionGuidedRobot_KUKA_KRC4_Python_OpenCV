## การกำหนดค่าการเชื่อมต่อ
กำหนดผ่านไฟล์ XML
C:\KRC\ROBOTER\Config\User\Common\EthernetKRL
- สามารถสร้างไฟล์เพิ่มเติมได้หากมีหลากหลาย Connection
- ทดลองทำการทดสอบ XML ผ่าน TCP/IP
### การใช้งานพอร์ต
เลือกพอร์ต Ethernet ให้ใช้งานได้แตกต่างกันตามสเปคของระบบ
- INTERFACE X66 (1 ช่อง)
- INTERFACE X67.1 - X67.3 (3 ช่อง)

### โครงสร้าง Program แบ่งออกเป็น 3 ส่วน
- ไฟล์ XML ให้ความสำคัญกับตัวพิมพ์เล็กและพิมพ์ใหญ่
- ชื่อของไฟล์ XML จะถูกใช้เป็น Access Key ใน KRL เช่น ถ้าตั้ง TESTKRL.XML >>> EKI_INIT("TESTKRL")

```xml
<ETHERNETKRL>

    <CONFIGURATION>
        <EXTERNAL>.....</EXTERNAL>
        <INTERNAL>.....</INTERNAL>
    </CONFIGURATION>
    
    <RECEIVE>
        <ELEMENTS>.....</ELEMENTS>
    </RECEIVE>
    
    <SEND>
        <ELEMENTS>.....</ELEMENTS>
    </SEND>
</ETHERNETKRL>

### Configuration
ส่วนสำหรับกำหนดค่าพารามิเตอร์การเชื่อมต่อระหว่างระบบภายนอกและ Interface
แบ่งออกเป็นสองส่วน
1. EXTERNAL
2. INTERNAL

```xml
    <CONFIGURATION>
        <EXTERNAL>.....</EXTERNAL>
        <INTERNAL>.....</INTERNAL>
    </CONFIGURATION>
```

### Data Reception
ส่วนกำหนดข้อมูลที่ Robot Controller จะรับ

```xml
    <RECEIVE>
        <ELEMENTS>.....</ELEMENTS>
    </RECEIVE>
```   

### Data Transmission
กำหนดโครงสร้างข้อมูลที่ Robot Controller จะส่ง

```xml
    <SEND>
        <ELEMENTS>.....</ELEMENTS>
    </SEND>
```    

### ERROR EKI00022
- Error while reading the configuration XML not valid. Originator <EthernerKRL>
#### สาเหตุ
- พิมพ์เล็กพิมพ์ใหญ่
- ลืมปิด Tag รึเปล่า เช่น <ETHERNETKRL> .... แล้วลืมปิด </ETHERNETKRL>
- วาง Element ถูกตำแหน่งไหม
- มี Tag ที่ EKI ไม่รู้จักรึเปล่า
- ไม่มี Xml Declaration
#### วิธีแก้ไข
- ทำการ Declare XML ไว้บนสุดของการโปรแกรม

หากยังไม่หายให้ทำการ
- ไล่เช็คแก้ตามข้อ 1 - 4 ให้ถูกต้อง

```xml
<?xml version="1.0" encoding="utf-8"?>
<ETHERNETKRL>
    <CONFIGURATION>
        <EXTERNAL>
            <IP>192.168.5.10</IP>
            <PORT>60000</PORT>
            <TYPE>Server</TYPE>
        </EXTERNAL>
        <INTERNAL>
            <ENVIRONMENT>Program</ENVIRONMENT>
            <BUFFERING Mode="FIFO" Limit="10"/>
            <BUFFSIZE Limit="16384"/>
            <TIMEOUT Connect="60000"/>
            <ALIVE Set_Out="666" Ping="200"/>
            <IP>192.168.5.1</IP>
            <PORT>54600</PORT>
            <PROTOCOL>TCP</PROTOCOL>
            <Messages Display="disabled" Logging="error"/>
        </INTERNAL>
    </CONFIGURATION>
</ETHERNETKRL>
``

### วิธีการดู LOG 
ผู้ใช้งานต้องอยู่ใน Expert Mode
1. ไปที่ Start up > Service > Minimize HMI
2. ไปที่ไฟล์ KUKA_DISC(C:) > KRC > ROBOTER > LOG > EthernetKRL
3. จะมีไฟล์ .csv แสดง log อยู่

### การค้นหา IP และ เชื่อมต่อกับ PC
#### Teach Pendant ต้องเป็น Expert Mode
1. ไปที่ Main Menu > Start Up > Network Configuration
2. ดูให้แถบด้านล่างเป็น KLI เราจะได้ IP ของ Robot นั้น

IP และ Subnet Mask ที่ได้จะแตกต่างกันออกไปตาม Robot นั้น ๆ

#### ตั้งค่าฝั่ง PC
- PC และ Robot จะต้องอยู่ใน Network เดียวกัน
1. ไปที่ Settings > Network & Internet > Ethernet
2. ไปตรงแถบ IP assignment แล้วคลิก Edit
3. ถ้า IP จาก Robot เป็น 192.168.5.1 ให้ตั้ง IP PC ให้อยู่ใน Subnet เดียวกัน เช่น 192.168.5.10 หลังจากนั้นให้ทดลองทำการเชื่อมต่อ

#### วิธีการทดลองการเชื่อมต่อ
1. เปิด CMD (Command Prompt)
2. รันคำสั่ง ping ไปที่ IP ของ Robot เช่น `ping 192.168.5.1`

หากเชื่อมต่อสำเร็จจะแสดงผลลัพธ์ประมาณนี้
```

### ส่วนของ External จะมี 3 ส่วน
ถ้ากำหนดให้ระบบภายนอกเป็น Server และ KUKA เป็น Client เราจำเป็นต้องกำหนด IP ของระบบภายนอก
1. TYPE: สำหรับกำหนดว่าระบบจะสื่อสารกับภายนอกในโหมดไหน
    - Server - ระบบภายนอกทำงานเป็น Server
    - Client - ระบบภายนอกทำงานเป็น Client
2. IP
    - IP Address ของระบบภายนอกเมื่อกำหนด TYPE = Server
    - ถ้า TYPE = Client ค่าของ IP จะไม่ถูกนำมาใช้งาน
3. PORT
    - หมายเลข Port ของระบบภายนอกเมื่อกำหนด TYPE = Server
    - รองรับค่า 1 - 65,534
    - ถ้า TYPE = client ค่าของ Port จะไม่ถูกนำมาใช้งาน

```xml
<?xml version="1.0" encoding="utf-8"?>
<ETHERNETKRL>
    <CONFIGURATION>
        <EXTERNAL>
            <IP>192.168.5.10</IP>
            <TYPE>Server</TYPE>
            <PORT>60000</PORT>
        </EXTERNAL>
    </CONFIGURATION>
</ETHERNETKRL>
```

### ส่วนของ Internal จะมี 9 ส่วน (ไม่จำเป็นต้องใส่ครบทุกส่วน)

#### 1. ENVIRONMENT (Optional)
ใช้กำหนดว่าเหตุการณ์ใดจะทำให้ EKI ตัด connection โดยอัตโนมัติ

**1.1 Program (ค่าเริ่มต้น)**
ตัด connection เมื่อ Robot Interpreter เกิดเหตุการณ์:
- Reset Program: กด Reset บน SmartPAD
- Deselect Program: ยกเลิกการเลือก Program
- Reconfigure I/Os: มีการ Config I/O ใหม่

เหมาะกับ: connection ที่ผูกกับ program cycle เช่น ส่งผลลัพธ์ทีละ cycle แล้วจบ

**1.2 Submit**
ตัด connection เมื่อ Submit Interpreter เกิดเหตุการณ์:
- Cancel Submit Interpreter: หยุด Submit
- Reconfigure I/Os

เหมาะกับ: connection ที่ผูกกับ background ตลอด แม้ program หลักจะ Reset หรือ Deselect

**1.3 System**
ตัด connection เฉพาะเมื่อ Reconfigure I/Os

เหมาะกับ: connection ที่ต้องการความต่อเนื่องสูงสุด เช่น Safety monitoring

---

#### 2. BUFFERING (Optional)
การจัดการ Data Memory

**Mode** — วิธีอ่านข้อมูลจาก Buffer
- FIFO (First In First Out) — ค่าเริ่มต้น เหมาะกับคำสั่งที่ต้องทำตามลำดับ เช่น Move A > Move B > Move C
- LIFO (Last In First Out) — อ่านค่าล่าสุดเสมอ

**Limit** — จำนวนข้อมูลสูงสุดที่เก็บใน Data Memory ได้
- รองรับค่า 1 - 512 (ค่าเริ่มต้น 16)

---

#### 3. BUFFSIZE (Optional)
**Limit** — จำนวน Byte สูงสุดที่รับได้โดยยังไม่ผ่านการแปลความหมายข้อมูล
- รองรับค่า 1 - 65,534 bytes (ค่าเริ่มต้น 16,384 bytes)

---

#### 4. TIMEOUT (Optional)
**Connect** — ระยะเวลาที่ใช้พยายามสร้างการเชื่อมต่อก่อนยกเลิก (หน่วยเป็น ms)
- รองรับค่า 0 - 65,534 ms (ค่าเริ่มต้น 2,000 ms)

---

#### 5. ALIVE (Optional)
ใช้ตรวจสอบสถานะการเชื่อมต่อ

**Set_Out** — ส่งสัญญาณออกไปข้างนอกหุ่นได้จริง เช่น ต่อสายไปหา PLC
- กำหนด Output เมื่อการเชื่อมต่อสำเร็จ, รองรับค่า 1 - 4,096
- Output จะถูก Set ตราบใดที่การเชื่อมต่อกับระบบภายนอกยังทำงานอยู่ และจะถูก Reset หากการเชื่อมต่อถูกตัดขาด

**Set_Flag** — ใช้แค่ภายใน KRL program เท่านั้น ไม่มีสัญญาณออกไปข้างนอก
- กำหนด Flag เมื่อการเชื่อมต่อสำเร็จ, รองรับค่า 1 - 1,025

**Ping** — ช่วงเวลาการส่ง Ping เพื่อตรวจสอบการเชื่อมต่อกับระบบภายนอก
- รองรับค่า 1 - 65,534 s

---

#### 6. IP*
- ใช้ได้เฉพาะเมื่อ `<EXTERNAL><TYPE>Client</TYPE></EXTERNAL>` (หุ่นเป็น Server รอรับ)

---

#### 7. Port*
- รองรับเฉพาะ 54,600 - 54,615 เท่านั้น (16 Channel)

---

#### 8. Protocol (Optional)
- TCP — ค่าเริ่มต้น แนะนำให้ใช้เสมอ
- UDP

---

#### 9. Message (Optional)

**Display** — การแสดงข้อความบน smartHMI
- disable — ปิดการแสดงข้อความ โดย Program จะจัดการเอง เหมาะกับ Production
- error — เปิดการแสดงข้อความตามปกติ (ค่าเริ่มต้น) แสดงเป็น Pop-up บน SmartPad เหมาะกับ Debug + Development

**Logging** — การบันทึกข้อความลงใน EKI Logbook
- warning — บันทึกทั้งข้อความแจ้งเตือนและข้อผิดพลาด
- error — บันทึกเฉพาะข้อความข้อผิดพลาด (ค่าเริ่มต้น)
- disable — ไม่บันทึกข้อความใด ๆ เลย


```xml
<?xml version="1.0" encoding="utf-8"?>
<ETHERNETKRL>
    <CONFIGURATION>
        <INTERNAL>
            <ENVIRONMENT>Program</ENVIRONMENT>
            <BUFFERING Mode="FIFO" Limit="10"/>
            <BUFFSIZE Limit="16384"/>
            <TIMEOUT Connect="60000"/>
            <ALIVE Set_Out="666" Ping="200"/>
            <IP>192.168.5.1</IP>
            <PORT>54600</PORT>
            <PROTOCOL>TCP</PROTOCOL>
            <Messages Display="disabled" Logging="error"/>
        </INTERNAL>
    </CONFIGURATION>
</ETHERNETKRL>
```
